```python
import pandas as pd
```

```python
df_fin = pd.read_csv("/Users/shang/downloads/数据分析/财务/四国联赛财务收支汇总.csv")
```

```python
key_cols = ['支出(万欧元)', '收入(万欧元)', '联赛']
df_clean = df_fin.dropna(subset=key_cols)
```

```python
df_clean['支出(万欧元)'] = pd.to_numeric(df_clean['支出(万欧元)'], errors='coerce').fillna(0)
df_clean['收入(万欧元)'] = pd.to_numeric(df_clean['收入(万欧元)'], errors='coerce').fillna(0)
```

```python
league_total = df_clean.groupby('联赛')[['支出(万欧元)', '收入(万欧元)']].sum().round(2)
print("各国联赛总收支（万欧）：")
print(league_total)
```

```python
import matplotlib.pyplot as plt
```

```python
league_total.index = ["CHINA", "GERMANY", "JAPAN", "ENGLAND"]
```

```python
leagues = league_total.index.tolist()
total_spend = league_total['支出(万欧元)'].tolist()   
total_income = league_total['收入(万欧元)'].tolist()  
bar_width = 0.35
x = range(len(leagues))
```

```python
plt.figure(figsize=(10, 6))
```

```python
plt.bar([i - bar_width/2 for i in x], total_spend, 
        width=bar_width, label='EXPEND.', color='orange')
```

```python
plt.bar([i + bar_width/2 for i in x], total_income, 
        width=bar_width, label='REVEN.', color='blue')
```

```python
for i in range(len(leagues)):
    plt.text(i - bar_width/2, total_spend[i] + total_spend[i]*0.02, 
             f'{total_spend[i]:.0f}', ha='center', fontsize=9)
    plt.text(i + bar_width/2, total_income[i] + total_income[i]*0.02, 
             f'{total_income[i]:.0f}', ha='center', fontsize=9)
```

```python
plt.title('Finance in Diferent Country', fontsize=14)
plt.xlabel('LEAGUE', fontsize=12)
plt.ylabel('NUMBER', fontsize=12)
plt.xticks(x, leagues) 
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7) 
```

```python
plt.savefig('0各国联赛收支对比柱状图.png', dpi=300, bbox_inches='tight')
plt.close()
```
