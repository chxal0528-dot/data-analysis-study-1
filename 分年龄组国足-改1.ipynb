```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.transfermarkt.com/chinese-super-league/dauerbrenner/wettbewerb/CSL/plus//galerie/0?saison_id=2024&pos=&detailpos=&altersklasse=o30"
```

```python
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", class_="items")
player_data = []

if table:
    for row in table.find_all("tr")[1:]:
        cells = row.find_all("td")
        if len(cells) >= 8: 
            name = cells[1].find("a").text.strip()  
            nation = cells[2].find("img")["title"] if cells[2].find("img") else "未知" 
            birth_age = cells[3].text.strip()  
            appearances = cells[5].text.strip()  
            play_minutes = cells[7].text.strip() 
            
            player_data.append({
                "姓名": name,
                "国籍": nation,
                "生日/年龄": birth_age,
                "出场数": appearances,
                "上场时长(分钟)": play_minutes
            })
df = pd.DataFrame(player_data).drop_duplicates(subset="姓名")
df.to_csv("中超30+球员_精准字段.csv", index=False, encoding="utf-8-sig")
```

```python
position_keywords = [
    "Attacking Midfield", "Right-Back", "Centre-Back", "Goalkeeper",
    "Defensive Midfield", "Central Midfield", "Centre-Forward",
    "Left-Back", "Right Winger", "Left Winger"]
```

```python
def clean_player_data(raw_data, age_group):
    cleaned_data = []
    for item in raw_data:
        cleaned_item = item.copy()
        cleaned_item["姓名"] = cleaned_item.pop("姓名及位置")
        name_with_pos = cleaned_item["姓名"]
        for pos in position_keywords:
            if pos in name_with_pos:
                cleaned_item["姓名"] = name_with_pos.replace(pos, "").strip()
                break
        
        cleaned_item["年龄分组"] = age_group
        
        cleaned_data.append(cleaned_item)
    return cleaned_data
```

```python
cleaned_23_30 = clean_player_data(age_23_30_data, "23-30岁")
cleaned_u23= clean_player_data(u23_data, "U23")
```
