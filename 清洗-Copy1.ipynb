```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_players = pd.read_csv("/Users/shang/downloads/数据分析/球员数据.csv") 
df_eastasia = pd.read_csv("/Users/shang/downloads/数据分析/中国男足2025东亚杯名单_姓名生日.csv")
df_finance = pd.read_csv("/Users/shang/downloads/数据分析/财务/四国联赛财务收支汇总.csv")
```

```python
df_players = df_players.drop_duplicates()
df_finance = df_finance.drop_duplicates() 
```

```python
# 2.2 缺失值填充（PPT3-49-51页 fillna）
# 数值型字段用均值填充，分类字段用默认值填充
df_players['出场数'] = pd.to_numeric(df_players['出场数'], errors='coerce')  # PPT3-53页 to_numeric
df_players['上场时长'] = pd.to_numeric(df_players['上场时长'], errors='coerce')
df_players['出场数'] = df_players['出场数'].fillna(df_players['出场数'].mean())  # 均值填充
df_players['上场时长'] = df_players['上场时长'].fillna(df_players['上场时长'].mean())
df_finance['支出(万欧元)'] = df_finance['支出(万欧元)'].fillna(0)  # 标量填充
df_finance['收入(万欧元)'] = df_finance['收入(万欧元)'].fillna(0)
```

```python
# 2.4 东亚杯名单年龄计算（基础数值运算，PPT2-22页 ndarray运算）
df_eastasia['出生年份'] = df_eastasia['生日'].str[:4].astype(int)
df_eastasia['年龄'] = 2025 - df_eastasia['出生年份']
```

```python
def convert_money(money):
    try:
        money_str = str(money)
        if '亿' in money_str:
            return float(money_str.replace('亿', '')) * 10000
        return float(money_str)
    except:
        return 0
df_finance['总支出_万欧元'] = df_finance['支出(万欧元)'].apply(convert_money)
```

```python
position_map = {
    'Goalkeeper': '门将', 'Left-Back': '后卫', 'Right-Back': '后卫', 'Centre-Back': '后卫',
    'Defensive Midfield': '中场', 'Central Midfield': '中场', 'Attacking Midfield': '中场',
    'Left Winger': '中场', 'Right Winger': '中场', 'Centre-Forward': '前锋'
}
def get_simple_position(pos_str)
    for key, value in position_map.items():
        if key in pos_str:
            return value
    return '其他'
df_players['简化位置'] = df_players['姓名及位置'].apply(get_simple_position)
```

```python
def age_group(age):
    if age <= 23:
        return 'Under 23'
    elif age <= 30:
        return '23-30'
    else:
        return 'Over 30'
```

```python
df_eastasia['年龄分组'] = df_eastasia['年龄'].apply(age_group)
```

```python
age_count = df_eastasia['年龄分组'].value_counts()
```

```python
plt.figure(figsize=(8, 8))
```

```python
age_count.plot(
    kind='pie',
    colors=['green', 'blue', 'orange'],
    explode=[0.1, 0],
    autopct='%1.1f%%',
    title='CHINA Football Age COUNT'
)
```
