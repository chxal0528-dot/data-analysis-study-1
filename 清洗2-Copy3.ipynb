```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_players = pd.read_csv("/Users/shang/downloads/数据分析/球员数据.csv") 
```

```python
df_players['年龄'] = df_players['生日/年龄'].str.extract(r'\((\d+)\)').astype(int)
df_golden = df_players[(df_players['年龄'] >=23) & (df_players['年龄'] <=30)]
```

```python
nation_avg = df_golden[df_golden['国籍'].isin(['中国', '日本', '英国', '德国'])].groupby('国籍')[['出场数', '上场时长']].mean().round(2)
```

```python
plt.figure(figsize=(10, 6))
nation_avg.plot(
    kind='bar',
    color=['green', 'blue'],
    grid=True,
    title='23-30 Avg.Apperance Time(s)Internat.',
    legend=['Times', 'Timelong'] )
plt.xlabel('Nation')
plt.ylabel('Avg(times;h)')
plt.savefig('3_02黄金年龄球员效率条形图.png', dpi=300)
```

```python
plt.close()
```

```python
df_china = df_players[df_players['国籍'] == '中国'].copy()
```

```python
def get_position(pos_str):
    pos = pos_str.split()[-1]
    if pos in ['Goalkeeper']:
        return 'GK'
    elif pos in ['Left-Back', 'Right-Back', 'Centre-Back']:
        return 'B'
    elif pos in ['Defensive', 'Central', 'Attacking', 'Left', 'Right']:
        return 'C'
    elif pos in ['Centre-Forward']:
        return 'F'
    else:
        return 'O'
df_china['位置'] = df_china['姓名及位置'].apply(get_position)
```

```python
position_count = df_china['位置'].value_counts()
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.figure(figsize=(10, 6))
position_count.plot(
    kind='barh',
    color=['red', 'lightblue', 'lightyellow', 'pink'],
    grid=True,
    title='Diferent Hundle in CHINA'
)
plt.xlabel('num.')
plt.ylabel('position')
plt.savefig('4_0国足位置分布条形图.png', dpi=300)
plt.close()
```

```python
pos_efficiency = df_china.groupby('位置')[['出场数', '上场时长']].mean().round(2)
```

```python
age_bins = [20, 23, 26, 29, 32, 35]  # 年龄区间
age_labels = ['20-22', '23-25', '26-28', '29-31', '32-34']
df_china['年龄区间'] = pd.cut(df_china['年龄'], bins=age_bins, labels=age_labels, right=False)
age_dist = df_china['年龄区间'].value_counts().sort_index()
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 12))
```

```python
pos_efficiency.plot(
    kind='bar',
    ax=ax1,
    color=['orange', 'blue'],
    width=0.6,
    legend=['Appearances', 'Playing Time']
)
ax1.set_title('Efficiency in Diferent Pos. in CHINA', fontsize=12)
ax1.set_xlabel('Position', fontsize=10)
ax1.set_ylabel('Avg.(times;h)', fontsize=10)
ax1.grid(axis='y', linestyle='--', alpha=0.7)
ax1.legend(loc='upper right', fontsize=9)
```

```python
age_dist.plot(
    kind='bar',
    ax=ax2,
    color='green',
    width=0.6
)
ax2.set_title('Age Count', fontsize=12)
ax2.set_xlabel('Age Group', fontsize=10)
ax2.set_ylabel('number', fontsize=10)
ax2.grid(axis='y', linestyle='--', alpha=0.7)
```

```python
plt.tight_layout()
```

```python
plt.savefig('00中国俱乐部本土球员复合图.png', dpi=300, bbox_inches='tight')
plt.close()
```

```python
df_finance = pd.read_csv("/Users/shang/downloads/数据分析/财务/四国联赛财务收支汇总.csv")
```

```python
def convert_money(money_str):
    if '亿' in str(money_str):
        return float(str(money_str).replace('亿', '')) * 10000
    elif '万' in str(money_str):
        return float(str(money_str).replace('万', ''))
    else:
        return float(money_str)
df_finance['支出(万欧元)'] = df_finance['支出(万欧元)'].apply(convert_money)
```

```python
df_league_fin = df_finance[df_finance['联赛'].isin(['中超', '日本J1联赛', '英超', '德甲'])].copy()
```

```python
plt.figure(figsize=(10, 6))
```

```python
df_league_fin.plot(
    x='联赛',
    y='支出(万欧元)',
    kind='bar',
    color=['orange', 'blue', 'green', 'yellow'],
    grid=True,
    title='Expenditure in Diferent Nations'
)
plt.xlabel('League')
plt.ylabel('Expenditure')
plt.savefig('5_0各国联赛支出柱状图.png')
plt.close()
```
