```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.transfermarkt.com/chinese-super-league/dauerbrenner/wettbewerb/CSL/plus//galerie/0?saison_id=2024&pos=&detailpos=&altersklasse=o30"
```

```python
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", class_="items")
player_data = []

if table:
    # 遍历表格行（跳过表头）
    for row in table.find_all("tr")[1:]:
        cells = row.find_all("td")
        if len(cells) >= 8:  # 确保列数足够
            # 精准匹配每一列的字段：
            name = cells[1].find("a").text.strip()  # 姓名（第2列）
            nation = cells[2].find("img")["title"] if cells[2].find("img") else "未知"  # 国籍（第3列，国旗标题）
            birth_age = cells[3].text.strip()  # 生日/年龄（第4列）
            appearances = cells[5].text.strip()  # 出场数（第6列）
            play_minutes = cells[7].text.strip()  # 上场时长（第8列）
            
            player_data.append({
                "姓名": name,
                "国籍": nation,
                "生日/年龄": birth_age,
                "出场数": appearances,
                "上场时长(分钟)": play_minutes
            })

# 3. 导出CSV（去重+整理）
df = pd.DataFrame(player_data).drop_duplicates(subset="姓名")  # 去重重复姓名
df.to_csv("中超30+球员_精准字段.csv", index=False, encoding="utf-8-sig")

# 4. 结果
print(f"✅ 爬取完成！共{len(df)}条有效数据")
print("📊 字段包含：姓名、国籍、生日/年龄、出场数、上场时长")
print("数据预览：")
print(df[["姓名", "国籍", "出场数"]].head(10))
```
